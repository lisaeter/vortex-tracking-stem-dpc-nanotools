```python
import hyperspy.api as hs
import pyxem as pxm
import matplotlib.pyplot as plt
from skimage import data
from skimage.filters import threshold_otsu
```

# Plot single data

```python
s = hs.load("data/2.zspy", lazy=True)
s_bs = s.get_direct_beam_position(method="center_of_mass")
s_bs.compute()
s_bs_lp = s_bs.get_linear_plane()
s_bs_corr = s_bs-s_bs_lp
```

```python
s_sum = s.sum(axis=(-1, -2))
s_sum.compute()
```

```python
s_sum.plot()
```

`mask` er signal objekt så kan også bare gjøre mask.plot()

```python
thresh = threshold_otsu(s_sum.data)
mask = s_sum > thresh
```

```python
fig, ax = plt.subplots()
ax.imshow(mask)
```
