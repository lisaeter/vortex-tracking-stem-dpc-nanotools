```python
import hyperspy.api as hs
import pyxem as pxm
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.anchored_artists import AnchoredSizeBar
import matplotlib.font_manager as fm
import matplotlib.patheffects as patheffects
import os
```

```python
def plot_s_bs(s_bs):
    fig, ax = plt.subplots()
    ax_indicator = ax.inset_axes([0.035, 0.06, 0.2, 0.2])
    pxm.utils.plotting.plot_beam_shift_color(s_bs, ax=ax, indicator_rotation=135, ax_indicator=ax_indicator, autolim_sigma=3.5)
    fig.tight_layout()
    return fig, ax

def add_drawings(ax):
    fontprops = fm.FontProperties(size=16)
    scalebar_kwargs = {'size': 5, 'label': '5 um', 'loc': 3, 'frameon': False, 'color': 'white', 'size_vertical': 0.4, 'label_top': False, 'fontproperties': fontprops}
    scalebar0 = AnchoredSizeBar(transform=ax.transData, **scalebar_kwargs)
    # Denne legger til et svart omriss rundt scalebar teksten, for å gjøre den lettere å lese
    scalebar0.txt_label._text.set_path_effects([patheffects.withStroke(linewidth=2, foreground='black', capstyle="round")])
    ax.add_artist(scalebar0)
    path_effects = [patheffects.withStroke(linewidth=2, foreground='black', capstyle="round")]
    ax.annotate("a", (0.025, 0.95), xycoords="axes fraction", fontsize=16, color="white", path_effects=path_effects)
```

This does the heavy lifting

```python
def get_s_bs_corr(file):
    s = hs.load(file, lazy=True)
    s_bs = s.get_direct_beam_position(method="center_of_mass")
    s_bs.compute()
    s_bs_lp = s_bs.get_linear_plane()
    s_bs_corr = s_bs-s_bs_lp
    return s_bs_corr
```

# Plot single data

```python
s_bs_corr = get_s_bs_corr("data/2.zspy")
fig, ax = plot_s_bs(s_bs_corr)
add_drawings(ax)
fig.savefig("2.png", dpi=300, bbox_inches="tight", pad_inches=0)
```

# Plot all data

```python
image_folder = "images/"
data_folder = "data/"

if not os.path.isdir(image_folder):
    os.mkdir(image_folder)

for i in range(len(os.listdir(data_folder))):
    file = f"{data_folder}{i}.zspy"
    s_bs_corr = get_s_bs_corr(file)
    fig, ax = plot_s_bs(s_bs_corr)
    add_drawings(ax)
    fig.savefig(f"{image_folder}{i}.png", dpi=300, bbox_inches="tight", pad_inches=0)
```
